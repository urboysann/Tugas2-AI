### Langkah 2: Fuzzifikasi

#### 2.1 Derajat Keanggotaan Pengalaman Kerja (x = 20)

Fungsi keanggotaan:
- **Sedikit** (turun linear dari 5 ke 15):
$$\mu_{Sedikit}(x) = \begin{cases} 1 & x \leq 5 \\ \frac{15-x}{15-5} & 5 < x < 15 \\ 0 & x \geq 15 \end{cases}$$

- **Banyak** (naik linear dari 10 ke 20):
$$\mu_{Banyak}(x) = \begin{cases} 0 & x \leq 10 \\ \frac{x-10}{20-10} & 10 < x < 20 \\ 1 & x \geq 20 \end{cases}$$

Perhitungan untuk x = 20:

$$\mu_{Sedikit}(20) = 0 \quad \text{(karena } x \geq 15\text{)}$$

$$\mu_{Banyak}(20) = 1 \quad \text{(karena } x \geq 20\text{)}$$

| Himpunan | Nilai Keanggotaan |
|---|---|
| Sedikit | **0.0000** |
| Banyak | **1.0000** |


#### 2.2 Derajat Keanggotaan Nilai Ujian (x = 7.0)

Fungsi keanggotaan:
- **Rendah** (turun linear dari 0 ke 5):
$$\mu_{Rendah}(x) = \begin{cases} 1 & x \leq 0 \\ \frac{5-x}{5-0} & 0 < x < 5 \\ 0 & x \geq 5 \end{cases}$$

- **Sedang** (segitiga, naik dari 2 ke 4, puncak di 4-6, turun dari 6 ke 8):
$$\mu_{Sedang}(x) = \begin{cases} 0 & x \leq 2 \ atau \ x \geq 8 \\ \frac{x-2}{4-2} & 2 < x < 4 \\ 1 & 4 \leq x \leq 6 \\ \frac{8-x}{8-6} & 6 < x < 8 \end{cases}$$

- **Tinggi** (naik linear dari 5 ke 10):
$$\mu_{Tinggi}(x) = \begin{cases} 0 & x \leq 5 \\ \frac{x-5}{10-5} & 5 < x < 10 \\ 1 & x \geq 10 \end{cases}$$

Perhitungan untuk x = 7.0:

$$\mu_{Rendah}(7) = 0 \quad \text{(karena } x \geq 5\text{)}$$

$$\mu_{Sedang}(7) = \frac{8 - 7}{8 - 6} = \frac{1}{2} = 0.5 \quad \text{(karena } 6 < x < 8\text{)}$$

$$\mu_{Tinggi}(7) = \frac{7 - 5}{10 - 5} = \frac{2}{5} = 0.4 \quad \text{(karena } 5 < x < 10\text{)}$$

| Himpunan | Nilai Keanggotaan |
|---|---|
| Rendah | **0.0000** |
| Sedang | **0.5000** |
| Tinggi | **0.4000** |

### Langkah 4: Inferensi Tsukamoto

Operator yang digunakan: **AND = min()**

Fungsi keanggotaan output Gaji:
- **Kecil** (turun linear dari Rp1jt ke Rp20jt):
$$\mu_{Kecil}(z) = \frac{20.000.000 - z}{20.000.000 - 1.000.000}$$
Inversnya: $z_{Kecil} = 20.000.000 - \alpha \times 19.000.000$

- **Besar** (naik linear dari Rp1jt ke Rp20jt):
$$\mu_{Besar}(z) = \frac{z - 1.000.000}{20.000.000 - 1.000.000}$$
Inversnya: $z_{Besar} = 1.000.000 + \alpha \times 19.000.000$

---

#### Rule 1: Sedikit AND Rendah → Kecil
$$\alpha_1 = min(\mu_{Sedikit}(20),\ \mu_{Rendah}(7)) = min(0.0000,\ 0.0000) = \mathbf{0.0000}$$
$$z_1 = 20.000.000 - 0.0 \times 19.000.000 = \mathbf{Rp\ 20.000.000}$$

#### Rule 2: Sedikit AND Sedang → Kecil

#### Rule 3: Sedikit AND Tinggi → Kecil

#### Rule 4: Banyak AND Rendah → Kecil

#### Rule 5: Banyak AND Sedang → Besar ✅

### Langkah 5: Defuzzifikasi (Metode Rata-rata Berbobot / Weighted Average)

Formula defuzzifikasi Tsukamoto:

$$z^* = \frac{\sum_{i=1}^{n} \alpha_i \cdot z_i}{\sum_{i=1}^{n} \alpha_i}$$

Substitusi nilai:

$$\text{Pembilang} = (\alpha_1 \cdot z_1) + (\alpha_2 \cdot z_2) + (\alpha_3 \cdot z_3) + (\alpha_4 \cdot z_4) + (\alpha_5 \cdot z_5) + (\alpha_6 \cdot z_6)$$

$$= (0 \times 20.000.000) + (0 \times 20.000.000) + (0 \times 20.000.000) + (0 \times 20.000.000)$$
$$+ (0.5 \times 10.500.000) + (0.4 \times 8.600.000)$$

$$= 0 + 0 + 0 + 0 + 5.250.000 + 3.440.000 = \mathbf{8.690.000}$$

$$\text{Penyebut} = \alpha_1 + \alpha_2 + \alpha_3 + \alpha_4 + \alpha_5 + \alpha_6$$
$$= 0 + 0 + 0 + 0 + 0.5 + 0.4 = \mathbf{0.9}$$

$$z^* = \frac{8.690.000}{0.9} = \mathbf{Rp\ 9.655.555{,}56}$$
